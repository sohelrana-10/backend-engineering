```python
import sqlite3 as sql

connection = sql.connect(r"D:\\backend-engineering\\Database\\practice.db")
cursor = connection.cursor()
```

```python
cursor.execute("SELECT * FROM students")

rows = cursor.fetchall()

for i in range(5):
    print(rows[i])
```

```python
cursor.execute("SELECT DISTINCT first_name  FROM students")
distinct_names=cursor.fetchall()

print(distinct_names)
```

```python
# cursor.execute("SELECT first_name FROM students  LIMIT 5")
cursor.execute("SELECT DISTINCT first_name FROM students  LIMIT 5 OFFSET 5")
cursor.execute("SELECT DISTINCT first_name FROM students  LIMIT 5,3")
top_five=cursor.fetchall()

print(top_five)
```

```python
cursor.execute(""" 
SELECT * 
FROM students
WHERE department= 'CSE'
LIMIT 5 OFFSET 5
""")

all_student=cursor.fetchall()
for student in all_student:
    print(student)
```

```python
cursor.execute("""

SELECT *FROM students
ORDER BY MARKS,AGE DESC
"""
)
top_students=cursor.fetchall()
for student in top_students:
    print(student)
```

```python
cursor.execute("""

SELECT *FROM students
ORDER BY 2,3 DESC
"""
)
top_students=cursor.fetchall()
for student in top_students:
    print(student)
```

```python
cursor.execute("""
SELECT *
FROM students
WHERE NOT department='CSE'
ORDER BY marks DESC
""")
for student in cursor.fetchall():
    print(student)
```

```python
cursor.execute("""
SELECT *
FROM students
WHERE first_name LIKE ('S%') AND department='CSE'
ORDER BY marks DESC
""")
for student in cursor.fetchall():
    print(student)
```
