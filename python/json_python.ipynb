```python
import json

patients = [
    {"id": "P001", "name": "Sohel", "age": 25},
    {"id": "P002", "name": "Rahim", "age": 30},
    {"id": "P003", "name": "Karim", "age": 22}
]

with open("patients.json", "w") as file:
    json.dump(patients, file, indent=4)

with open("patients.json", "r") as file:
    data = json.load(file)

for patient in data:
    print(patient["id"], patient["name"], patient["age"])
```

```python
import json
import os

FILE_NAME = "patients.json"


def load_patients():
    """Read existing patients from the JSON file."""
    if not os.path.exists(FILE_NAME):
        return []

    with open(FILE_NAME, "r") as file:
        return json.load(file)


def save_patients(patients):
    """Save all patients to the JSON file."""
    with open(FILE_NAME, "w") as file:
        json.dump(patients, file, indent=4)


def add_patient():
    """Add a new patient."""
    patients = load_patients()

    patient_id = input("Enter patient ID: ").strip()

    # Prevent duplicate patient IDs
    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient ID already exists!")
            return

    name = input("Enter patient name: ").strip()
    age = int(input("Enter patient age: "))

    new_patient = {
        "id": patient_id,
        "name": name,
        "age": age
    }

    patients.append(new_patient)
    save_patients(patients)

    print("Patient added successfully!")


def display_patients():
    """Display all patients."""
    patients = load_patients()

    if not patients:
        print("No patients found.")
        return

    for patient in patients:
        print(
            f"ID: {patient['id']} | "
            f"Name: {patient['name']} | "
            f"Age: {patient['age']}"
        )


def search_patient():
    """Search for a patient by ID."""
    patients = load_patients()

    patient_id = input("Enter patient ID to search: ").strip()

    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient found:")
            print(f"ID: {patient['id']}")
            print(f"Name: {patient['name']}")
            print(f"Age: {patient['age']}")
            return

    print("Patient not found.")


def main():
    while True:
        print("\n--- Patient Management System ---")
        print("1. Add patient")
        print("2. Display all patients")
        print("3. Search patient")
        print("4. Exit")

        choice = input("Choose an option: ").strip()

        if choice == "1":
            add_patient()
        elif choice == "2":
            display_patients()
        elif choice == "3":
            search_patient()
        elif choice == "4":
            print("Goodbye!")
            break
        else:
            print("Invalid choice. Try again.")


if __name__ == "__main__":
    main()
```
